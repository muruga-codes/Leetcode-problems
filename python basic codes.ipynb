write a python program to identify wheather a number is a strong number or not

strong number is the sum of the factorial of its digit is equal to the original number

ex:145

1!+4!+5!

```python
def strong_number(s):
    if s<=0:
        return
    if not s:
        return
    original=s
    total=0
    while s>0:
        digit=s%10
        fact=1
        for i in range(1,digit+1):
            fact*=i
        total+=fact
        s//=10
    if total==original:
        print("strong number")
    else:
        print("not a strong number")
s=int(input("s"))
strong_number(s)
        
```

write a program for Binary to octal conversion 

i/p: 101011

o/p: 53

explanation : 101011=> divide into two parts 101 011 then 101=>5 011=>3 the output 53

```python
def binary_octal(binary):
    if binary=="":
        print("invalid input")
        return
    elif any(ch not in "01" for ch in binary):
        print("invalid binary number")
        return
    else:
        decimal=0
        for digit in binary:
            decimal=decimal*2+int(digit)
        if decimal==0:
            print(0)

        else:
            octal=""
            while decimal>0:
                remainder=decimal%8
                octal=str(remainder)+octal
                decimal//=8
            print(octal)
binary=input().strip()
binary_octal(binary)
```

```python
def toggle(s):
    if not s:
        return
    if s=="":
        print("invalid input")
        return
    result=""
    for ch in s:
        if ch.isupper():
            result+=ch.lower()
        elif ch.islower():
            result+=ch.upper()
        else:#this case for number,special character,space they cannot change
            result+=ch
    print(result)
s=input().strip()
toggle(s)
```

write a program to count common subsequence in two string

i/p: s1=ABCDEF  S2=AEDFHR

o/p: 3

explanation : The longest common subsequence is ADF,so its count/length is 3

```python
def common_subsequence(s1,s2):
    if not s1 and not s2:
        return
    if s1=="" and s2=="":
        print("invalid input")
        return
    m=len(s1)
    n=len(s2)
    dp=[[0]*(n+1) for _ in range(m+1)]
    for i in range(1,m+1):
        for j in range(1,n+1):
            if s1[i-1]==s2[j-1]:
                dp[i][j]=dp[i-1][j-1]+1
            else:
                dp[i][j]=max(dp[i-1][j],dp[i][j-1])
    print(dp[m][n])
s1=input().strip()
s2=input().strip()
common_subsequence(s1,s2)
```

```python
def scalar_product(a,b):
    if len(a)==0 and len(b)==0:
        print("empty vector")
        return
    if len(a)!=len(b):
        print("both have same size")
        return
    a.sort()
    b.sort(reverse=True)
    result=0
    for i in range(len(a)):
        result+=a[i]*b[i]
    return result
a=list(map(int, input().split()))
b=list(map(int, input().split()))
print(scalar_product(a,b))
```

write a program to find whether arrays are disjoint or not

i/p: 

arr1=1 2 3 4

arr2=5 6 7 8

o/p: Arrays are disjoint

Disjoint=two arrays have no common element

```python
def disjoint_array(arr1,arr2):
    if len(arr1)==0 or len(arr2)==0:
        print("the array is empty")
        return
    disjoint=True
    for i in range(len(arr1)):
        for j in range(len(arr2)):
            if arr1[i]==arr2[j]:#if two array has common element the disjoint will false 
                disjoint=False
                break#stop the inner loop
        if disjoint==False:
            break #stop the outer loop

    if disjoint:
        print("Array are disjoint")
    else:
        print("not disjoint it has common element")

arr1=list(map(int,input().split()))
arr2=list(map(int,input().split()))
disjoint_array(arr1,arr2)
```

```python
def array_equal(arr):
    if not arr:
        return
    if any(x<=0 for x in arr):
        print("only postive numbers are allowed")
        return

    for i in range(len(arr)):
        while arr[i]%2==0:
            arr[i]//=2
        while arr[i]%3==0:
            arr[i]//=3
        while arr[i]%5==0:
            arr[i]//=5
    if all(x==arr[0] for x in arr):
        print("Yes")
    else:
        print("No")
arr=list(map(int,input("arr:").split()))
array_equal(arr)
```
