Write a Python program to find the sum of digits of a given integer.

```python
def sum_digit(s):
    if not s:
        return
    n=0
    while s>0:
        n+=s%10
        s//=10
    print(n)
s=1234
sum_digit(s)
```

Write a Python program to reverse a given integer.

```python
def reverse_integer(s):
    if not s:
        return
    n=0
    while s>0:
        n=n*10+s%10
        s//=10
    print(n)
s=1234
reverse_integer(s)
```

write a python program to find a lcm of two numbers

```python
def lcms(a,b):
    if not a and not b:
        return 
    if a>b:
        lcm=a
    else:
        lcm=b
    while True: 
        if lcm%a==0 and lcm%b==0:
            break
        lcm+=1
    print("LCM:",lcm)
a=int(input("a"))
b=int(input("b"))
lcms(a,b)

```

write a python program to identify wheather a number is a strong number or not

strong number is the sum of the factorial of its digit is equal to the original number

ex:145

1!+4!+5!

=1+24+120=

145 is a strong number

```python
def strong_number(s):
    if s<=0:
        return
    if not s:
        return
    original=s
    total=0
    while s>0:
        digit=s%10
        fact=1
        for i in range(1,digit+1):
            fact*=i
        total+=fact
        s//=10
    if total==original:
        print("strong number")
    else:
        print("not a strong number")
s=int(input("s"))
strong_number(s)
        
```

write a python program to find GCD of two numbers

i/p:

a=48

b=18

o/p:

6

```python
def gcd(a,b):
    if a==0 and b==0:
        print("gcd not defined")
        return
    while b!=0: #loop stop when b==0
        a,b=b,a%b
    print("GCD:",a)
a=int(input("a"))
b=int(input("b"))
gcd(a,b)
```

write a python program to check wheather the given number is prime or not

o/p: Prime number

i/p: 1

o/p: Not a prime number

```python
def prime_number(n):
    if not n:
        return
    if n<=1:
        return "not a prime number"
    for i in range(2,n):
        if n%i==0:
            return "not a prime number"
    return "prime number"
n=int(input("n:"))
print(prime_number(n))
```

write a python program to find the factorial of a given number

i/p: 5

o/p: 120

```python
def factorial(n):
    if not n:
        return 
    if n<0:
        return
    fact=1
    while n>0:
        fact*=n
        n-=1
    return fact
n=int(input("n:"))
print(factorial(n))
```

```python
def fact(n):
    if not n:
        return 
    if n<0:
        return
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
n=int(input("n:"))
print(fact(n))
```
