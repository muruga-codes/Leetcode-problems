```python
def vowel_constant(s):
    if s is None:
        return None
    ch="aeiouAEIOU"
    if len(s)!=1 or not s.isalpha():
        print("enter a single aplphebet")

    elif s in ch:
        print("vowels")
    else:
        print("consonant")
s=input("enter s:")
vowel_constant(s)
```

Write a Python program to identify whether a given character is an alphabet or not.

```python
def alphabet_not(s):
    if s is None:
        return None
    if len(s)==1 and s.isalpha():#why because it check character
        print("alphabet")
    else:
        print("not a alphabet")
s=input("enter s:")
alphabet_not(s)
```

Write a Python program to find the ASCII value of a character.

Character → ASCII = ord()

ASCII → Character = chr()

```python
def asciis(s):
    if s is None:
        return None
    if len(s)==1:#character
        print(ord(s))
s=input("s:")
asciis(s)
```

```python
def digits(s):
    if not s:
        return 
    s=12345
    count=0
    n=str(abs(s)).replace("."," ")
    for i in range(n):
        count+=1
    print(count)
digits(s)

```

Write a Python program to find the sum of digits of a given integer.

```python
def sum_digit(s):
    if not s:
        return
    n=0
    while s>0:
        n+=s%10
        s//=10
    print(n)
s=1234
sum_digit(s)
```
