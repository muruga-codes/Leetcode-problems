write a python program to identify wheather a number is a strong number or not

strong number is the sum of the factorial of its digit is equal to the original number

ex:145

1!+4!+5!

=1+24+120=

145 is a strong number

```python
def strong_number(s):
    if s<=0:
        return
    if not s:
        return
    original=s
    total=0
    while s>0:
        digit=s%10
        fact=1
        for i in range(1,digit+1):
            fact*=i
        total+=fact
        s//=10
    if total==original:
        print("strong number")
    else:
        print("not a strong number")
s=int(input("s"))
strong_number(s)
        
```

write a python program to identify if the number is perfect number or not

perfect number means the whole number that equals the sum of its (+) divisor

i/p: 6

o/p: 6 expalination : divisor are 1,2,3 so sum=1+2+3=6

```python
def perfect_number(n):
    if not n:
        return
    if n<=0:
        print("give positive number")
        return 
    original=n
    sums=0
    for i in range(1,n):
        if n%i==0:
            sums+=i
    if sums==original:
        print("perfect number")
    else:
        print("not a perfect number")
n=int(input("n:"))
perfect_number(n)
```

write a python program to identify if the number is armstrong number or not 

```python
def armstrong(n):
    if not n:
        return
    if n<=0:
         print("give positive number")
         return 
    num=n
    total=0
    power=len(str(n))
    while n>0:
        digit=n%10
        total+=digit**power
        n//=10

    if total==num:
        print("armstrong")
    else:
        print("not armstrong")
n=int(input("n:"))
armstrong(n)
```

A prime number is a number that is greater than 1 that is divisible by only 1 and itself

```python
def primenumber_range(start,end):
    if start>end:
        return 
    for n in range(start,end+1):
        if n<2:
            continue # skip the number that is less than 2 like 0,1,2
        is_prime=True #initially assume n is prime number

        for i in range(2,n):
            if n%i==0:
                is_prime=False #if the given number has more than 2 factor that is not a prime number then it is false
                break
        if is_prime:
            print(n,end=" ")
start=int(input("start:"))
end=int(input("end:"))
primenumber_range(start,end)    
```

write a python program to print armstrong number between two intervals

i/p:

start:100

end:500

o/p:

153 370 371 407

```python
def armstrong_two_interval(start,end):
    if start>end:
        return
    found=False
    for n in range(start,end+1):
        if n<0:
            continue
        original=n
        total=0
        power=len(str(n))
        while n>0:
            digit=n%10
            total+=digit**power
            n//=10
        if original==0:
            total=0 #edge case for 0

        if total==original:
            print(original,end=" ")
            found=True

    if not found:
        print("armstron not in this range")
start=int(input("start:"))
end=int(input("end:"))
armstrong_two_interval(start,end)

```

```python
def is_prime(n):#this function check the number is prime or not
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True

def prime_number(n):#this function calculate the sum of 2 prime number
    if n<4: #the two sum if 2+2=4 so less 4 is not a prime number 
        print("cannot be expressed as sum of two prime number")
        return
    found=False
    for i in range(2,n//2+1):#why divide because it remove the repeation like 10=3+7 and 10=7+3 it will print only one 
        if is_prime(i) and is_prime(n-i):
            print(n, "=", i, "+", n-i)
            found=True
    if not found:                                                                                               
        print("no pair of prime number found")
n=int(input("n:"))
prime_number(n)
        
```

write a python program to print pyramid pattern using stars

```python
def pyramid(n):
    if n<0:
        return
    for i in range(n):
        for j in range(i,n):
            print(" ",end=" ")
        for j in range(i):
            print("*",end=" ")
        for j in range(i+1):
            print("*",end=" ")
        print()
n=int(input("n:"))
pyramid(n)
    
```
